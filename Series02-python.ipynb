# IEAP Python Series 02: Sinusoidal Signals

This notebook is about how to generate and visualize a sinusoidal signal using Python. The goal is to move from the mathematical representation of a sine wave to its implementation and visualization using NumPy and Matplotlib.

# Math to generate a sine wave
$$ x(t) = A sin(\omega t + \phi) $$

with:

* $A$ is the amplitude of the signal
* $\omega$ is the angular frequency (rad/s)
* $\phi$ is the phase of the signal



```python
# Import the needed libraries
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Create a numpy array called t that contains 1000 values from 0 to 1 (t is for time)
t = np.linspace(0, 1, 1000)

# Create a numpy array called s that contains the values of a sine function of t (s is for signal)
s = np.sin(2 * np.pi * 1 * t)  # 1 Hz sine wave with a phase shift zero

# Plot the signal
plt.plot(t, s)
plt.xlabel('Time (s)')
plt.ylabel('Amplitude')
plt.title('Sine Wave')
plt.grid(True)
plt.show()

# save a plot in png file
plt.savefig('sine_wave.png')
```

# Sinusoidal signals (better version)

```python
# Create the cosine signal
c = np.cos(2 * np.pi * 1 * t)  # 1 Hz cosine wave with a phase shift zero

# Create a figure with two y-axes
fig, ax1 = plt.subplots(figsize=(10, 5))

# Reuse the sine signal from the previous cell
line1 = ax1.plot(t, s, label="Sine wave")
ax1.set_xlabel("Time (s)")
ax1.set_ylabel("Sine amplitude")
ax1.grid(True)

# Create the second y-axis
ax2 = ax1.twinx()

# Add the cosine signal
line2 = ax2.plot(t, c, label="Cosine wave", color='orange')
ax2.set_ylabel("Cosine amplitude")

# Add title
ax1.set_title("Sine and Cosine Waves")

# Add combined legend
lines = line1 + line2
labels = [line.get_label() for line in lines]
ax1.legend(lines, labels, loc="upper right")

# Save as PDF
fig.tight_layout()
fig.savefig("sine_cosine_waves.pdf")

# Display the graph
plt.show()
```

# Sinusoidal Signal (Final Version)
A reusable function is created to generate sinusoidal signals with different frequencies.

```python
# create the function to generate the sine waves
def sinusoid(t, amplitude, frequency, phase):
    """
    Generate a sinusoidal signal.
    
    Parameters:
    t (numpy array): Time values
    amplitude: Amplitude of the signal
    frequency: Frequency of the signal in Hz
    phase: Phase of the signal in radians
    
    Returns:
    numpy array containing the sinusoidal signal
    """
    return amplitude * np.sin(
        2 * np.pi * frequency * t + phase
    )
```

```python
# Create the time array 
t = np.linspace(0, 1, 1000)

# Define signal parameters
amplitude = 1
phase = 0
frequency1 = 1
frequency2 = 5

# Generate two signals with different frequencies
s1 = sinusoid(t, amplitude, frequency1, phase)
s2 = sinusoid(t, amplitude, frequency2, phase)
```

```python
# Create the figure
fig, ax = plt.subplots(figsize=(10, 5))

# Plot the two sinusoidal signals
ax.plot(t, s1, label="Sine wave (1 Hz)")
ax.plot(t, s2, label="Sine wave (5 Hz)")

# Add title and labels
ax.set_title("Comparison of Sinusoidal Signals")
ax.set_xlabel("Time (s)")
ax.set_ylabel("Amplitude")

# Add grid and legend
ax.grid(True)
ax.legend()

# Display the graph
plt.show()
```
